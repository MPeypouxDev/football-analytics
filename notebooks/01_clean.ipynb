```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("../data/raw/database.sqlite")
player_attributes = pd.read_sql_query("SELECT * FROM Player_Attributes", conn)
player_attributes.head()
```

```python
player_attributes.shape
print(player_attributes.shape)
player_attributes.dtypes
print(player_attributes.dtypes)
```

```python
player_attributes["date"] = pd.to_datetime(player_attributes["date"])
player_attributes.dtypes["date"]
```

```python
player_attributes["year"] = player_attributes["date"].dt.year
(player_attributes["year"] == 2007).sum()
```

```python
player_attributes_clean = player_attributes[player_attributes["year"] >= 2008]
player_attributes_clean.shape
```

```python
player_attributes_clean["overall_rating"].isna().sum()
```

```python
player_attributes_final = player_attributes_clean.dropna(subset=["overall_rating"])
player_attributes_final.shape
```

```python
player_attributes_final.to_csv("../data/processed/player_attributes_clean.csv", index=False)
```
